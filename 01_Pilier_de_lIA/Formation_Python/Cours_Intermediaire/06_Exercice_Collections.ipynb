<h1>Exercice : Analyse de Dataset avec Collections</h1>

Objectifs à accomplir :

1. Utilisez Counter pour :

- Compter le nombre de transactions par catégorie
- Identifier les 3 produits les plus vendus
- Analyser la répartition des types de clients (Premium/Standard)

2. Utilisez defaultdict pour :

- Regrouper les transactions par date
- Calculer le chiffre d'affaires total par catégorie
- Créer un mapping client_type → liste des produits achetés

3. Utilisez namedtuple pour :

- Créer une structure Transaction avec les champs appropriés
- Recréer le dataset avec cette structure

4. Analyse finale :

- Calculez le panier moyen par type de client
- Identifiez le jour avec le plus gros chiffre d'affaires

```python
# Provided data
transactions = [
    ("2023-01-15", "Électronique", "Smartphone", 599.99, "Premium"),
    ("2023-01-15", "Vêtements", "T-shirt", 19.99, "Standard"),
    ("2023-01-16", "Électronique", "Laptop", 899.99, "Premium"),
    ("2023-01-16", "Électronique", "Smartphone", 599.99, "Standard"),
    ("2023-01-17", "Alimentation", "Café", 4.50, "Standard"),
    ("2023-01-17", "Vêtements", "Jean", 49.99, "Premium"),
    ("2023-01-18", "Électronique", "Smartphone", 599.99, "Premium"),
    ("2023-01-18", "Alimentation", "Thé", 3.25, "Standard"),
]
```

<h2>Setup</h2>

```python
# Standard libraries
from collections import Counter, defaultdict, namedtuple
```

<h2>Part One</h2>

-> `Counter()`

```python
# Count the number of transaction by product 
count_by_category = Counter([transaction[1] for transaction in transactions])

# Display the result
print(count_by_category)
```

```python
# Found the 3 items the most celled 
most_celled = Counter([transaction[2] for transaction in transactions]).most_common(3)

# Display the result
print(most_celled)
```

```python
# Count the client types 
count_client_type = Counter([transaction[-1] for transaction in transactions])

# Display results
print(count_client_type)
```

<h2>Part Two</h2> 

-> `defaultdict()`

```python
# Instantiate a defaultdict object 
groupes = defaultdict(list)

# Group transactions by date
for transaction in transactions:
    groupes[transaction[0]].append(transaction[1:])

# Display result 
for item in groupes.items():
    print(item)

print(type(groupes))
```

```python
# Instantiate a defaultdict object to build a counter
revenue_by_category = defaultdict(float)

for transaction in transactions:
    revenue_by_category[transaction[1]] += transaction[3]

# Display result
for cat in revenue_by_category.items():
    print(cat)
```

```python
# Instantiate a defaultdict object to build a mapping
product_by_client_type = defaultdict(list)

for transaction in transactions:
    product_by_client_type[transaction[-1]].append(transaction[2])

# Displat result
for client_type in product_by_client_type.items():
    print(client_type)
```

<h2>Part Three</h2>

-> `namedtuple()`

```python
# Instantiate a `Transaction` structure
Transaction = namedtuple("Transaction", ["date", "category", "product", "price", "client_type"])

# Instantiate an empty list to rebuild initial dataset
transactions_struct = []

# Rebuild the initial dataset using that structure 
for date, category, product, price, client_type in transactions:
    transactions_struct.append(Transaction(date, category, product, price, client_type))

# Display result
print(transactions_struct)
```

<h2>Part Four</h2>

<h3><u>Compute the average shopping cart</u></h3>

```python
client_type_shopping_cart_lists = defaultdict(list)

for transaction in transactions_struct:
    client_type_shopping_cart_lists[transaction.client_type].append(transaction.price)


average_shopping_cart = defaultdict(float)

for client_type in client_type_shopping_cart_lists.items():
    #print(client_type[0])
    average_shopping_cart[f"{client_type[0]}"]= sum(client_type[1]) / len(client_type[1])

# Display the result
average_shopping_cart
```

```python
# Another possible solution 
# Calculer le panier moyen par type de client
panier_moyen = defaultdict(float)
nombre_transactions = defaultdict(int)

for transaction in transactions_struct:
    panier_moyen[transaction.client_type] += transaction.price
    nombre_transactions[transaction.client_type] += 1

for client_type, total in panier_moyen.items():
    panier_moyen[client_type] = total / nombre_transactions[client_type]

print("Panier moyen par type de client:")
for client_type, montant in panier_moyen.items():
    print(f"{client_type}: {montant:.2f} €")
```

<h3><u>Identify the day with the bigger revenue</u></h3>

```python
# Instantiate a defaultdict object to group transactions by date
group_by_day = defaultdict(float)

for transaction in transactions_struct:
    group_by_day[transaction.date] += transaction.price

# Display result
print(group_by_day)                     # Display the whole defaultdict object
for date in group_by_day.items():       # Display each item extracted from tje defaultdict object
    print(date)

print(group_by_day.values())            # Display only the values from the defaultdict
print(group_by_day['2023-01-15'])       # Display only a specific value using the key

# Extract the key whose value is the biggest
best_day_revenue_key = max(group_by_day, key=lambda k: group_by_day[k]) 
print(f"\nThe date with the biggest revenue is: {best_day_revenue_key}\n")

# Get the biggest value 
print(group_by_day[best_day_revenue_key])


# Another way to do it exactly equaly 
best_day_revenue = max(group_by_day, key=group_by_day.get)
print(best_day_revenue)
```
