<h1>Exercice : Optimisation d'Hyperparamètres avec itertools</h1>

Vous travaillez sur un projet de machine learning et devez tester différentes combinaisons d'hyperparamètres pour optimiser votre modèle. Utilisez les outils d'itertools pour générer et analyser ces combinaisons.

# Contexte :

Vous avez un modèle avec les hyperparamètres suivants :

- learning_rate : [0.01, 0.1, 0.2]
- batch_size : [16, 32, 64]
- epochs : [10, 20, 50]
- optimizers : ["Adam", "SGD"]

# Objectifs à accomplir :

1. Utilisez product pour :

- Générer toutes les combinaisons possibles d'hyperparamètres
- Calculer le nombre total de configurations à tester
- Afficher les 5 premières combinaisons

2. Utilisez combinations pour :

- Sélectionner 3 combinaisons parmi les 5 premières pour un test rapide
- Créer des paires de configurations pour les comparer

3. Utilisez permutations pour :

- Générer différents ordres de test des optimizers
- Créer des séquences d'entraînement avec différents learning_rates

4. Utilisez chain et islice pour :

- Combiner différents groupes de configurations
- Limiter le nombre de tests à effectuer (ex: les 10 premières)

<h2>Setup</h2>

```python
# Standard libraries
import random
from itertools import product, combinations, permutations, chain, islice, combinations_with_replacement
```

```python
# Définition des hyperparamètres
learning_rates = [0.01, 0.1, 0.2]
batch_sizes = [16, 32, 64]
epochs = [10, 20, 50]
optimizers = ["Adam", "SGD"]
```

<h2>Part One</h2>

-> `product()`

```python
# Create a list with all hyperperameter
all_HP = [learning_rates, batch_sizes, epochs, optimizers]
```

```python
# 1. Generate all possible hyperparameters combinations
all_config = list(product(*all_HP))
# Display results
print(all_config)
```

On peut, soit mettre tous les liste dans une table et utiliser l'opérateur `*`, ou alors, simplement mettre chaque liste séparément comme paramètre de la fonction `product`.

```python
list(product(learning_rates, batch_sizes, epochs, optimizers))
```

```python
# 2. Compute the total number of configuration to test
len(all_config)
```

```python
# 3. Display the 5 first combinations
first_5 = all_config[:5]

# Display result
print(first_5)
```

<h2>Part Two</h2>

-> `combinations()`

```python
# 1. Select the first combination of 3 HP's different combination sets
extract_3_combination_sets = list(combinations(first_5, 3))[0]

# Display results
print(extract_3_combination_sets)
```

```python
# 2. Generate pairs of those combination sets to compare them
list(combinations(extract_3_combination_sets, 2))
```

```python
# 3. Generate pairs from `all_config` to compare them
all_config_pairs = list(combinations(all_config, 2))

# Display infos 
print(all_config_pairs[:6])
len(all_config_pairs) # 1431 pairs to allow to compare each HP's combination set 
```
