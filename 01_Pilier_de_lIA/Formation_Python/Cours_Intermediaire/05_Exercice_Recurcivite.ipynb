```python
# FIRST TRIAL

def update_employee_position(employee_name: str
                             , new_position: str
                             , json_data: dict  
                             , current_branch: dict = {}
                             , current_level: int = 0
                             , current_key: str=""
                             , current_value: str=""):
    """ Define a function that use recursivity to look for an employee and update it's position
    
        Args : 
            employee_name (str): the name of the employee to find
            new_position (str): the new position to replace
            json_data (dict): the initial json containt
            current_branch (dict): the data part we are looking in 
            current_level (int): the current deep level reached
            current_key (str): the current dictionary key explored
            current_value (str): the current dictionnaire key's value explored
    """
    print(f"x--[CURRENT LEVEL]: {current_level}")


    # IF .. current level is the ground level, then copy the initial data into to current branch
    if current_level == 0:
        current_branch = json_data

    print(f"[CURRENT BRANCH]: {current_branch}")

    # Extract the keys within a list
    current_keys_list = list(current_branch.keys())
    print(f"\n[KEYS]: {current_keys_list}\n")

    # Iterate through each key 
    for idx, key in enumerate(current_keys_list):
        # Update current key and values
        current_key = key
        current_value = current_branch[key]

        # Help debugging 
        print(f"[CURRENT KEY]: {current_key}")
        print(f"[CURRENT VALUE]: {current_value}")


         # IF .. the value is the name we are looking for, then stop de process
        if current_value == employee_name:
            # Update the employee position 
            #current_branch[current_key] = new_position
            current_branch[current_keys_list[idx+1]] = new_position
            #print(f"\n[YOU GOT IT]\n")
            #print(f"\n[DATA]:\n{json_data}\n")
    
            # Return the path to reach the name we were looking at
            return  json_data #(json_data, current_path)



        # IF .. the current branch key's value is "str" type, this means we have reach a "leaf" then continue
        if isinstance(current_value, str) or (isinstance(current_value, list) and isinstance(current_value[0], str)):
            print(f"--[LEAF REACHED]\n")
            continue
        # ELIF .. the current branch key's value is "list" type, and its elements are "dict"s 
        # .. this means we can go deeper within the structure
        # .. then dive inside
        elif isinstance(current_value, list) and isinstance(current_value[0], dict):
            # Update the branche level 
            current_level += 1
            # Iterate through each dict contains on the list 
            for branch in current_value :
                # Update the current branch
                new_current_branch = branch
                print(f"[NEW CURRENT BRANCH: {new_current_branch}\n")

                # Call itself to repeat te protocol untill to find the employee we are looking for
                update_employee_position(employee_name, new_position, json_data, new_current_branch, current_level, current_key, current_value)

        
```

<h3><u>Correction</u></h3>

```python
def modifier_role(obj, nom_cible, nouveau_role):
    """
    Recherche récursive dans un JSON pour trouver une personne par son nom
    et modifier son rôle.
    """
    # IF .. the input object is a "dict" then ..
    if isinstance(obj, dict):
        # IF .. the object "nom" key is the one we are looking for then ..
        if obj.get("nom") == nom_cible:
            # IF .. the object has a "role" attribute
            if "rôle" in obj:
                print(f"✓ Trouvé {nom_cible} avec le rôle : {obj['rôle']}")
                # Update the "role" attribut  
                obj["rôle"] = nouveau_role
                print(f"→ Rôle modifié en : {nouveau_role}")
                
        # Iterate through each item from tje dict 
        for key, value in obj.items():
            # Call itself 
            modifier_role(value, nom_cible, nouveau_role)

    # ELIF .. the input object is a "list" then .. 
    elif isinstance(obj, list):

        # Iterate through each element from the list
        for item in obj:
            modifier_role(item, nom_cible, nouveau_role)

```
