<h1>Ordinal Encoding</h1>

<h2>Setup</h2>

```python
# Standard libraries
import numpy as np
import pandas as pd
import seaborn as sns

# Preprocessing libraries
from sklearn.preprocessing import OrdinalEncoder
```

<h2>Diamond Exemple</h2>

```python
# Data path 
Diamonds_data_path = "/Users/johanndurand-vegas/GIT/AI-Practical_ML-PRO/01_Pilier_de_lIA/Methodologie_IA/4_Preprocessing/data/diamonds.csv"
```

```python
# Load the diamond dataset into a DataFrame
#df = sns.load_dataset("diamonds") # ERROR : certificate pb
df = pd.read_csv(Diamonds_data_path)

# Display the table
df.head()
```

<h3>1. <u>Encodage brutale</u></h3>

```python
# Instantiate an encoder object from the OrdinalEncoder class
encoder = OrdinalEncoder()

# Adjust the encoder to our data : the three columns
encoder.fit(df[["cut", "color", "clarity"]])

# Transform the categories of each column into numerical values
encoder.transform(df[["cut", "color", "clarity"]])
```

<h3>2. <u>Définir l'ordre de l'encodage</u></h3>

```python
for col in ['cut', 'color', 'clarity']:
    print(f"{col} : {df[col].unique()}")
```

```python
# Order to respect 
cut_order = ['Fair', 'Good', 'Very Good', 'Premium', 'Ideal']
color_order = ['J', 'I', 'H', 'G', 'F', 'E', 'D']
clarity_order = ['I1', 'SI2', 'SI1', 'VS2', 'VS1', 'VVS2', 'VVS1', 'IF']
```

```python
# Instantiate an encoder object from the OrdinalEncoder class
encoder = OrdinalEncoder(categories=[cut_order, color_order, clarity_order])

# Adjust the encoder to our data : the three columns
encoder.fit(df[["cut", "color", "clarity"]])

# Transform the categories of each column into numerical values
encoder.transform(df[["cut", "color", "clarity"]])
```

<h3>3. <u>Cas de nouvelles catégories</u></h3>

```python
# Instantiate an encoder object from the OrdinalEncoder class
encoder = OrdinalEncoder(
			categories=[cut_order, color_order, clarity_order]
			, handle_unknown = 'use_encoded_value'
			, unknown_value = -1
		)

# Adjust the encoder to our data : the three columns
encoder.fit(df[["cut", "color", "clarity"]])

# Transform the categories of each column into numerical values
encoder.transform(df[["cut", "color", "clarity"]])
```

```python
# Try with a new value
encoder.transform([["Fair", "G", "A+"]])
```
