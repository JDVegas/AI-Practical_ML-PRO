<h2>Fonction Sigmoïde</h2>

La fonction sigmoid est très souvent retrouvé en data science. Il est très important de la maîtriser. 

- Rappeler la formule de la fonction sigmoid et l'écrire dans une fonction python
- Calcule sigmoid(0)
- Tracer la fonction sigmoid dans l'intervale [-5,5]
- Démontrer que les limites de la fonction sigmoid sont 0 et 1
- Calculer la valeur de sigmoid(10000) et sigmoid(-10000) ? Que se passe-t-il ?

<h3><u>Formule fonction sigmoid</u></h3>

$f(x)\ =\ {1\over{1+e^{-x}}}$

```python
# Def sigmoid
def sigmoid(x):
    return 1 / (1+np.exp(-x))
```

```python
# Test the funciton 
sigmoid(0)
```

```python
# Build a space with 100 values from -5 to 5
X = np.linspace(-5,5, 100)

# Compute y 
y = sigmoid(X)
```

```python
# Build the graphical representation
plt.scatter(X,y)

# Define coordinate
coord = 0

# Add lines to check values
plt.vlines(x=coord, ymin=0, ymax=sigmoid(coord), color='r')
plt.hlines(y=sigmoid(coord), xmin=X.min(), xmax=coord, color='r')

# Display the graph
plt.show()
```

Une fois qu'on a tracé les lignes horizontales et verticales à partir des valeurs de X et de y, eh bien on retombe bien sur le même résultat pour sigmoid(0). 

<u>Démontrer que les limites de la fonction sigmoid sont 0 et 1</u>:

$\lim\limits_{\begin{array}{l} x \to -\infty\end{array}}\ {1\over {1+e^{-x}}} =\ {\lim\limits_{\begin{array}{l} x \to -\infty\end{array}}\ 1 \over \lim\limits_{\begin{array}{l} x \to -\infty\end{array}}\ 1+e^{-x}}\ =\ {1 \over +\infty}\ =\ 0$

$\lim\limits_{\begin{array}{l} x \to +\infty\end{array}}\ {1\over {1+e^{-x}}} =\ {\lim\limits_{\begin{array}{l} x \to +\infty\end{array}}\ 1 \over \lim\limits_{\begin{array}{l} x \to +\infty\end{array}}\ 1+e^{-x}}\ =\ {1 \over {1+0}}\ =\ 1$

Si on souhaite pouvoir mieux observer ces limites, on peut générer le même graphique mais sur un intervalle plus grand

```python
# Build a space with 100 values from -5 to 5
X = np.linspace(-20,20, 100)

# Compute y 
y = sigmoid(X)
```

```python
# Build the graphical representation
plt.scatter(X,y)
plt.show()
```

Ainsi, on peut observer des plateaux aux niveaux des valeurs de 0 et de 1 qui représentent les limites de la fonction. 

```python
# Compute
print(sigmoid(10000))
print(sigmoid(-10000))
```

Pour -10000, on a de nouveau un message d'erreur avec un `overflow`. Cette valeur d'exponentielle est beaucoup trop grande à calculer. 

<h2>Fonction logit</h2>

Pour rappel, la fonction logit est la réciproque de la fonction sigmoid

- Rappeler la formule de la fonction `Logit`et écrire la fonction avec Python
- Calculer la valeur Logit(10). Que se passe-t-il ? 
- Tracer la fonction Logit sur son domaine de définition 
- Soit X = 2.24, y = sigmoid(X), z = logit(y), que vaut z ? 

<h3><u>Formule</u></h3>

$f(x) = log({x \over {1-x}})$

```python
# Define the logit function
def logit(x):
    return np.log(x/ (1-x))
```

```python
# Compute
logit(10)
```

On rencontre un nouveau message d'erreur. Pour rappel, quand on utilise une fonction, il faut avant tout connaître son domaine de définition. C'est à dire l'intervalle de valeurs que celle-ci peut prendre en entrée.

Or ici, on se trouve en présence d'une fonction logarithme. Et si on s'en rappelle bien, la fonction logarithme est définie sur tout l'intervalle positif, en excluant au passage la valeur de 0. Par conséquent, il faut que $x\over {1-x}$ soit TOUJOURS, supérieur à 0. 

Quel serait l'intervalle de $x$ qui permette d'avoir ${x\over {1-x}}>0$ ? 

La fonction logit peut exister que $\forall x \in ]0, 1[$
En dehors de cette intervalle, le quotient $x\over {1-x}$ ne donnera que des valeurs négatives. 
Raison pour laquelle la valeur 10 ne peut pas être calculé. Il sort du domaine de définition de la fonction.

<h3><u>Display x/(1-x)</u></h3>


```python
# Create a space with 100 values 
X = np.linspace(-0.5, 1.5, 100)
# Compute y
y = X / (1-X)

# Build the graph
plt.plot(X, y)
plt.show()
```

```python
# Build the graph
plt.plot(X, y)
plt.ylim(0)
plt.show()
```

Analysons cette courbe. Pour que l'image de la fonction $f(x)={x\over{1-x}}$ donne des valeurs positives, il faut que $x$ se trouve entre 0 et 1. 

C'est pouquoi, quand on utilise la fonctio logit, on lui <u>donne des probabilités</u> !!!  
Qu'on va écrire : $f(p)\ =\ {p\over {1-p}}$


```python
# Compute concistent values
print(logit(0.5))
print(logit(0.3))
```

```python
# Create a space with 100 values 
X = np.linspace(0.001, 0.999, 100)
# Compute y
y = logit(X)

# Build the graph
plt.scatter(X, y)
plt.grid()
plt.show()
```

```python
# Compute 
X = 2.24
y = sigmoid(X)
z = logit(y)

print(f"z value is :{z}")
```

z vaut x. Eh oui !! Ce qu'il faut bien comprendre, c'est que la fonction logit, et la fonction sigmoid sont des fonctions réciproques l'une de l'autre.
