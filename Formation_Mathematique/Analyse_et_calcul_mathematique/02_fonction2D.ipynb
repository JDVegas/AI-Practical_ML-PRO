<h2>SetUp</h2>

```python
# Standard libraries
import numpy as np
import matplotlib.pyplot as plt
```

<h2>Créer un plan</h2>

Pour rappel, un plan, c'est le résultat d'une fonction linéaire à 2 variables. $f(x_1, x_2)\ =\ w_1x_1\ +\ w_2x_2\ +\ b$ 

Dans la pratique, si on veut calculer et afficher un plan, il nous faut calculer les résultats de cette fonction pour toutes les valeurs $x_1$ et $x_2$ de notre espace. C'est-à-dire, calculer les valeurs pour chaque combinaisons du plan $f(0, 0), f(0, 1), f(1, 0), f(1, 1), f(0, 2), f(1, 2), f(2, 2), f(2, 1), f(2, 0), f(0,3), etc$, ce qui représente beaucoup d'oéprations. 

Pour réaliser cela, on va utiliser la fonction `np.meshgrid()` de la librairies Numpy. 

```python
# Define range of values for each variable
x1 = np.arange(-5, 5, 1)
x2 = np.arange(-5, 5, 1)
```

```python
# Generate the grids
x1_grid, x2_grid = np.meshgrid(x1, x2)

# Display grids
print(f"x1_grid:\n{x1_grid}\n")
print(f"x2_grid:\n{x2_grid}\n")
```

À l'aide de ces deux matrice, on peut ensuite effectuer des calcules afin de remplir toute notre matrice qui contiendra les éléments constituant notre plan. 

Pour cela, on va commencer par définir des paramètres $w_1, w_2, b$ pour notre modèle $f$, et construire notre modèle $f$. 

```python
# Define model parameters
w1 = 0
w2 = 0.5
b = 0

# Define model
f = w1 * x1_grid + w2 * x2_grid + b

# Display f matrix
print(f"f:\n{f}\n")
```

On retrouve bien ici dans la matrice résultante de la fonction $f$, les résultats du modèle pour chaque combinaison d'entrée $(x_1, x_2)$

```python
# Build the graphic
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the plan
ax.plot_surface(x1_grid, x2_grid, f, alpha=0.6, color="Blue", edgecolor="none")
```

On retrouve donc bien un plan dans un graphique avec des valeurs allant de 1 à 5. Bien entendu, le plan pourrait s'étendre jusqu'à l'infini. Ici on observe que le plan pour les valeurs de notre espace qu'on a calculé, mais si on augmente le nombre de valeurs alors le plan va s'agrandir.

<h2>Fonction MSE</h2>

Voyons maintenant comment tout cela s'articule dans un cas concret: celui d'une fonction MSE (Mean Squared Error).  Il s'agit d'une des fonctions coût qu'on utilise dans les problèmes de régression. 

```python
# Define a seed
np.random.seed(0)

# Define data points
X = np.linspace(0,10,20)

# Define true parameters
true_w, true_b = 2, 1

# Define true model
y = true_w * X + true_b + np.random.normal(scale=2, size=X.shape)

# Display the results of the function 
plt. scatter(X,y)
```

Ce qu'on obtient ici, c'est un ensemble de points qui pourraient représenter le prix de différentes maison en fonction d'une de leur caractértistiques. Ce qu'on peut faire donc, c'est de développer un petit modèle pour essayer de s'approcher le plus possible de ce comportement. 

Pour cela, on va commencer par fixer les paramètre $w$ et $b$ de manière aléatoire. Puis de les utiliser dans un modèle linairr. 

```python
# Arbitrary choice of w and b parameters
w, b = 0.7, 5

# Define the lineaire model using the input data previously defined
y_pred = w * X + b
```

```python
# Display the true model and our linear model on the graphic 
plt.scatter(X,y, color='blue')
plt.plot(X, y_pred, color="green")
```

Ce qu'il va nous intéresser en data science c'est de déterminer la distance qu'il y a entre notre modèle vert et le prix des vraies maisons (les points blues). 

Pour cela on va utiliser l'erreur quadratique moyenne (MSE) :
$MSE\ =\ {1\over m}\sum(y^{(i)}-\hat{y}^{(i)})^2$

Plus précisement, il s'agit de calculer l'erreur qu'il y aurait entre les points prédit de la droite en vert ($\hat{y}$) et les points bleus ($y$). C'est-à-dire, de mesurer l'écart / la distance (en la métant au carré afin qu'elle soit positive), et en faisant la somme pour tous les points de notre dataset. Enfin, on divise par le nombre de points ce qui va nous donner une moyenne.

```python
# Define MSE
mse = 1/len(y)*np.sum((y - y_pred)**2)
# Display MSE
mse
```

On peut se demander, quelles sont les variables de cette fonction MSE ? On peut se dire que la fonction MSE dépend de 2 entrées $MSE(y, \hat{y})$, donc $y$ et $\hat{y}$. Mais est-ce exacte ? Or, le prix des maisons $y$ ce n'est aps vraiment une variable qu'on peut modifier. En effet, ce sont des données réelles. Ce qu'on peut faire varier en revanche, ce sont les poinds du modèle permettant de calculer $\hat{y}$. C'est-à-dire, $MSE(w,b)$, où $w$ et $w$ sont les paramètres qu'on retrouve à l'intérieur de $\hat{y}$. 

Ce qui donne ainsi :$MSE\ =\ {1\over m}\sum(y^{(i)}-w*x^{(i)}+b)^2$

On retrouve ainsi nos deux variables. On peut dès lors visualiser l'allure de cette fonction par le biais de la méthode `meshgrid()` qu'on a pu voir plus haut. À noter que `meshgrid` peut travailler avec plus de 2 variables à la fois. 


```python
# Define a set of values for each variables 
W_vals = np.linspace(0,3,50)
B_vals = np.linspace(-1, 3, 50)

# Build the two matrices 
W, B = np.meshgrid(W_vals, B_vals)
```

Ainsi, grâce à `meshgrid` on obtient 2 tableaux qui vont nous permettre de calculer toutes les valeurs de notre MSE par rapport aux points de notre espace qu'on aura définis. 

```python
# Initialise MSE matrix with a 0 matrix
MSE = np.zeros_like(W)

# Compute the MSE for each value of the matrix
for i in range(W.shape[0]):
    for j in range(W.shape[1]):
        y_hat = W[i,j] * X + B[i, j]
        MSE[i, j] = np.mean((y- y_hat)**2)
```

```python
# Build the graphic
fig = plt.figure(figsize=(12, 8))

ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(W, B, MSE, cmap='viridis', edgecolor='none')

ax.set_xlabel('w')
ax.set_ylabel('b')
ax.set_zlabel('MSE')
ax.set_title("Surface de l'erreur quadratique moyenne (MSE)")

# Display the result
plt.show()
```

En traçant ces résultats on va pouvoir afficher une surface, et l'objectif en Machiner Learning, ce va être d'atteindre le minimum de cette surface. C'est-à-dire, le point le plus bas, car cela nous donnera le modèle qui fait les plus petites erreures. 

On rappel que ce qu'on visualise ici c'est la MSE. Donc plus on se trouve sur un point élevé de la courbe et plus les erreurs faites par notre modèle sont grandes.
