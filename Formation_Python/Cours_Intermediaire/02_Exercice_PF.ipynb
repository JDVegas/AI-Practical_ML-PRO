<h2>SetUp</h2>


```python
from functools import reduce
```

<h2>Data</h2>

```python
# Define data
sales_data = [
    {'product_id': 'A101', 'category': 'Electronics', 'price': 120.00, 'quantity': 3, 'region': 'North'},
    {'product_id': 'B205', 'category': 'Apparel', 'price': 45.99, 'quantity': 10, 'region': 'South'},
    {'product_id': 'A101', 'category': 'Electronics', 'price': 120.00, 'quantity': 5, 'region': 'West'},
    {'product_id': 'C302', 'category': 'Home Goods', 'price': 25.50, 'quantity': 8, 'region': 'North'},
    {'product_id': 'B205', 'category': 'Apparel', 'price': 45.99, 'quantity': 2, 'region': 'East'},
    {'product_id': 'D401', 'category': 'Electronics', 'price': 299.99, 'quantity': 1, 'region': 'West'},
    {'product_id': 'C302', 'category': 'Home Goods', 'price': 25.50, 'quantity': 12, 'region': 'South'},
    {'product_id': 'B205', 'category': 'Apparel', 'price': 45.99, 'quantity': 7, 'region': 'North'},
    {'product_id': 'A101', 'category': 'Electronics', 'price': 120.00, 'quantity': 2, 'region': 'East'}
]
```

<h2>Code</h2>

<h3><u>Question 1</u></h3>

```python
# Compute price ventilated by sail 
sails_prices_list = list(map(lambda x: x["price"]*x["quantity"],sales_data))
# Display
print(sails_prices_list)

# Compute total revenu 
total_revenu = round(reduce(lambda x,y: x+y, sails_prices_list), 2)
# Display
print(total_revenu)
```

<h3><u>Question 2</u></h3>

```python
# Filter out the sails that are above 300$
sails_filterd = list(filter(lambda x: x["price"]*x["quantity"] >300, sales_data))
# Display result
for sail in sails_filterd : 
    print(sail)
```

<h3><u>Question 3</u></h3>

```python
# Extract product ID list
product_ID_list = list(map(lambda x: str.lower(x['product_id']), sales_data))
# Display 
print(product_ID_list)
```

<h3><u>Question 4</u></h3>


```python
# Extract sails made on the North region
north_sails = list(filter(lambda x: x['region']=='North', sales_data))
# Display 
for sail in north_sails:
    print(sail)
```

```python
# Extract the list of quantities
quantities_list = list(map(lambda x: x['quantity'], north_sails))
# Display 
print(quantities_list)

# Compute the bigest quantity sailed
max_quantity = reduce(lambda x,y: x if x>y else y,  quantities_list)
# Display 
print(f"The max quantity sailed on the north region is : {max_quantity}")
```
