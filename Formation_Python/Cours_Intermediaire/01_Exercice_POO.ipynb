<h2>Setup</h2>

```python
from typing import List
```

<h2>Code</h2>

```python
# Define parent class
class Vehicule:
    # Define constructor 
    def __init__(self, marque:str, vitesse_max:float):
        # Define attributs 
        self._brand = marque
        self.__max_velocity = vitesse_max

    # Define methods
    # --x - o - x - o - x - o - x - o - x --
    # Def a function that will display that message when the Python object is printed
    def __str__(self) -> str:
        return f"\nThe vehicule is a {self.__class__.__name__} built by {self._brand}"

 
    def __eq__(self, autre: "Vehicule")->bool:
        # Two vehicule are equal if they have the same brand and the same velocity
        return self._brand == autre._brand and self.__max_velocity == autre.__max_velocity

    def move(self) -> str:
        return f"The vehicule {self._brand} is moving"
    # --x - o - x - o - x - o - x - o - x --
```

```python
# Define child class
class Car(Vehicule):
    # Define the constructor
    def __init__(self, marque:str, vitesse_max:float, portes:int):
        # Call for parent attributs and methods
        super().__init__(marque, vitesse_max)

        # Define more attributs 
        self.doors = portes

    def move(self) -> str:
        return f"The car {self._brand} is moving"
```

```python
# Define child class
class Motorbike(Vehicule):
    # Define constructor 
    def __init__(self, marque:str, vitesse_max:float):
        # Call for parent attributs and methods
        super().__init__(marque, vitesse_max)

    def move(self) -> str:
        return f"The motorbike {self._brand} is moving."
```

```python
# Define a list of vehicules creating 3 Python object using parent and child classes
vehicules: List[Vehicule] = [
    Vehicule("Airbus", 50.0)
    , Car("Peugeot", 180.0, 4)
    , Motorbike("Honda", 200.0)
]
```

```python
# Iterate through each vehicule 
for v in vehicules:
    # Check the readable presentation
    print(v)
    # Check de polymorphism
    print(v.move())
```

```python
# Create two cars with a different number of doors
v1 = Car("Toyota", 150.0, 4)
v2 = Car("Toyota", 150.0, 5)
```

```python
# Check if those cars are equal
print(v1==v2)
```
