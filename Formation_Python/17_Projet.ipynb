## Tâche 1 : Chargement des données

Nous allons créer une fonction qui charge le fichier `individus.json` (il se situe dans le dossier fichiers_projet) et retourne les données sous forme de liste de dictionnaires. Nous vérifierons également que le fichier existe.

On créera également une fonction `save_to_json` qui prendra en paramètre les données et le nom du fichier et les sauvegardera dans un fichier JSON.

```python
# Define data path 
input_data_path = "data/Projet_Debutant/individus.json"
os.path.exists(input_data_path)

# Output data Path
saved_file_path = "data/Projet_Debutant/individus_saved.json"
```

```python
def load_json(filename: str) -> list:
    """
    Charge un fichier JSON et retourne son contenu.

    Args:
        filename (str): Nom du fichier JSON.

    Returns:
        list ou dict: Contenu du fichier JSON, ou None si le fichier n'existe pas.
    """

    # Check that the file exists 
    if not os.path.exists(filename):
        return print(f"The file does not exists !")

    # Open the file 
    with open (filename, mode="r", encoding="utf-8") as file: 
        # Deserialise a json file into a dict
        individuous_dict = json.load(file)

    # Help debugging
    print(f"[individuous_dict] type is : {type(individuous_dict)}.")
    print(f"The list contains {len(individuous_dict)} elements.")
    print(f"[individuous_dict] first element type is :{type(individuous_dict[0])}")
    print(f"\nDisplay [individuous_dict]\n{individuous_dict}\n")

    return individuous_dict




# Charger les données initiales
#individus = load_json("fichiers_projet/individus.json")
individus = load_json(input_data_path)
if individus is not None:
    print(f"Données chargées avec succès. Nombre d'individus : {len(individus)}")
```

```python
def save_to_json(data, file_path):
    """
    Sauvegarde une structure de données dans un fichier JSON.

    Args:
        data: Données à sauvegarder (list, dict, etc.).
        filename (str): Nom du fichier JSON.
    """
    
    # Open the file 
    with open (file_path, mode="w", encoding="utf-8") as file: 
        # Serialise a json file into a dict
        json.dump(data, file, indent=2, ensure_ascii=True)

    # Help debugginh
    print(f"\nData saved:\n{data}\n")
    

# Save data into a file
save_to_json(individus, saved_file_path)

# Check that the file exists 
if os.path.exists(saved_file_path):
    print("The file exists ! Good Job.")
```

## Tâche 2 : Ajout du champ "taille"

Nous allons ajouter un champ `taille` (en mètres) à chaque individu, généré aléatoirement entre 1.50 et 2.00. Les données mises à jour seront sauvegardées dans un nouveau fichier JSON.

```python
def add_taille(individus):
    """
    Ajoute un champ 'taille' à chaque individu avec une valeur aléatoire.

    Args:
        individus (list): Liste de dictionnaires représentant les individus.

    Returns:
        list: Liste mise à jour avec le champ 'taille'.
    """
    # Iterate through each person from the list
    for indiv in individus:
        indiv['taille'] = round(random.uniform(1.5, 2.0), 2)

    # Help debugging 
    #print(f"List modified:\n{individus}\n")

    return individus

# Ajouter le champ taille et sauvegarder
individus_with_taille = add_taille(individus)

# Help debugging 
print(f"List modified:\n{individus_with_taille}\n")

# Save data
save_to_json(individus_with_taille, saved_file_path)
print("Champ 'taille' ajouté et sauvegardé.")
```

## Tâche 3 : Ajout du champ "IMC"

Nous allons calculer l'IMC pour chaque individu à l'aide de la formule : IMC = poids / (taille * taille). Le résultat sera arrondi à 2 décimales et ajouté comme champ `imc`. Les données mises à jour seront sauvegardées dans un nouveau fichier JSON.

```python
# Ajout du champ "IMC"

def calculate_imc(poids, taille):
    """
    Calcule l'IMC à partir du poids et de la taille.

    Args:
        poids (float): Poids en kilogrammes.
        taille (float): Taille en mètres.

    Returns:
        float: IMC arrondi à 2 décimales.
    """
    return round(poids / taille**2, 2)
    


def add_imc(individus):
    """
    Ajoute un champ 'imc' à chaque individu.

    Args:
        individus (list): Liste de dictionnaires représentant les individus.

    Returns:
        list: Liste mise à jour avec le champ 'imc'.
    """
    # Iterate through each person from the list
    for indiv in individus:
        indiv['imc'] = calculate_imc(indiv['poids'], indiv['taille'])

    return individus


# Ajouter le champ IMC et sauvegarder
individus_with_imc = add_imc(individus_with_taille)

# Help debugging 
print(f"List modified:\n{individus_with_taille}\n")

# Save data 
save_to_json(individus_with_imc, saved_file_path)
print("Champ 'imc' ajouté et sauvegardé.")
```

## Tâche 4 : Ajout du champ "pays de naissance"

Nous allons créer un dictionnaire associant chaque ville de naissance à un pays, puis ajouter un champ `pays_naissance` à chaque individu. Si une ville n'est pas dans le dictionnaire, le pays sera "Inconnu". Les données mises à jour seront sauvegardées dans un nouveau fichier JSON.

```python
# Create a dictionary city -> country
ville_pays = {
    "Paris": "France",
    "Lisbonne": "Portugal",
    "Berlin": "Allemagne",
    "Tokyo": "Japon",
    "Sydney": "Australie",
    "New York": "États-Unis",
    "Moscou": "Russie",
    "Rio": "Brésil",
    "Londres": "Royaume-Uni"
}
```

```python
def add_pays_naissance(individus, mapping):
    """
    Ajoute un champ 'pays_naissance' à chaque individu en fonction de la ville de naissance.

    Args:
        individus (list): Liste de dictionnaires représentant les individus.

    Returns:
        list: Liste mise à jour avec le champ 'pays_naissance'.
    """
    # Iterate trough each person
    for indiv in individus:
        # IF .. the ville_naissance exists on the mapping then ..
        if  indiv['ville_naissance'] in mapping:
            # .. we save the country within the profil 
            indiv['pays_naissance'] = mapping[indiv['ville_naissance']]
        # ELSE .. the ville_naissance does not exists, then ..
        else : 
            # .. the value is set to "Inconnu"
            indiv['pays_naissance'] = "Inconnu"

    return individus

# Ajouter le champ pays_naissance et sauvegarder
individus_with_pays = add_pays_naissance(individus_with_imc, ville_pays)

# Help debugging 
#print(f"List modified:\n{individus_with_pays}\n")

# Save results
save_to_json(individus_with_pays, saved_file_path)
print("Champ 'pays_naissance' ajouté et sauvegardé.")
```

## Tâche 5 : Analyse des âges

Nous allons calculer et afficher des statistiques sur les âges (moyen, minimum, maximum, nombre de personnes de plus de 50 ans). Les résultats seront sauvegardés dans un fichier JSON.

```python
# Analyse des âges

def analyze_ages(individus):
    """
    Calcule des statistiques sur les âges des individus.

    Args:
        individus (list): Liste de dictionnaires représentant les individus.

    Returns:
        dict: Dictionnaire contenant les statistiques sur les âges.
    """
    # Create a dictionary to store stat computed using ages
    stats_ages = dict()

    # Extract the ages of each person
    ages_list = [indiv['age'] for indiv in individus_with_pays]

    # Compute each statistique and store them into the new dictionary 
    # -- x - o - x  o - x  o - x  o - x  o - x --
    stats_ages['moyenne'] = sum(ages_list)/len(ages_list)
    #stats_ages['moyenne'] = math.fsum(ages_list)/len(ages_list)
    stats_ages['min'] = min(ages_list)
    stats_ages['max'] = max(ages_list)

    stats_ages['more_than_50'] = 0
    for val in ages_list:
        if val > 50:
            stats_ages['more_than_50'] += 1 


    # -- x - o - x  o - x  o - x  o - x  o - x --

    return stats_ages


# Analyser les âges et afficher les résultats
stats_ages = analyze_ages(individus_with_pays)
print("Statistiques sur les âges :")
for key, value in stats_ages.items():
    print(f"- {key.replace('_', ' ').capitalize()} : {value}")

# Sauvegarder les statistiques
save_to_json(stats_ages, "data/Projet_Debutant/stats_ages.json")
print("Statistiques sauvegardées dans 'stats_ages.json'.")
```

## Tâche 6 : Analyse des IMC

Nous allons identifier et afficher le nombre d'individus dans différentes catégories d'IMC (maigreur, poids normal, surpoids). Les résultats seront sauvegardés dans un fichier JSON.

```python
# Create a dict with the IMC limites
imc_dict = {
   "maigreur": (0, 18.4)
   , "poids normal": (18.5, 24.9)
   , "surpoids": (25,)
}
```

```python
def analyze_imc(individus, IMC_values):
    """
    Calcule des statistiques sur les IMC des individus.

    Args:
        individus (list): Liste de dictionnaires représentant les individus.

    Returns:
        dict: Dictionnaire contenant les statistiques sur les IMC.
    """
    # Create a dictionary to store statistics
    imc_statistics = {"maigreur": 0, "poids normal": 0, "surpoids": 0}

    # Iterate through each person
    for indiv in individus:
        # Help debugging
        #print(f"\nIMC: {indiv['imc']}\n")

        # IF .. 
        if indiv['imc'] <= IMC_values['maigreur'][1]:
            imc_statistics['maigreur'] += 1
        elif indiv['imc'] >= IMC_values['poids normal'][0] and indiv['imc'] <= IMC_values['poids normal'][1]:
            imc_statistics['poids normal'] += 1
        elif indiv['imc'] >= IMC_values['surpoids'][0]:
            imc_statistics['surpoids'] += 1

    return imc_statistics


# Analyser les IMC et afficher les résultats
stats_IMC = analyze_imc(individus_with_pays, imc_dict)
print("Statistiques sur les IMC :")
for key, value in stats_IMC.items():
    print(f"- {key.replace('_', ' ').capitalize()} : {value}")

# Sauvegarder les statistiques
save_to_json(stats_IMC, "data/Projet_Debutant/stats_IMC.json")
print("Statistiques sauvegardées dans 'stats_IMC.json'.")
```
